```python
import numpy as np


def bayes_update(p_prior, p_alarm_given_hazard=0.90, p_alarm_given_safe=0.10):
  p_alarm = (p_alarm_given_hazard * p_prior) + (
      p_alarm_given_safe * (1 - p_prior)
  )
  return (p_alarm_given_hazard * p_prior) / p_alarm


prior_values = [0.05, 0.20, 0.50, 0.80]
for p_h in prior_values:
  res = bayes_update(p_h)
  print(f"P(H) = {p_h:.2f} -> P(H|ALARM) = {res:.4f}")


def hmm_filter(observations):
  p_h, p_s = 0.20, 0.80
  for t, obs in enumerate(observations, 1):
    p_h_pred = p_h * 0.80 + p_s * 0.10
    p_s_pred = 1 - p_h_pred
    l_h, l_s = (0.90, 0.10) if obs == "ALARM" else (0.10, 0.90)
    unnorm_h = p_h_pred * l_h
    unnorm_s = p_s_pred * l_s
    norm_sum = unnorm_h + unnorm_s
    p_h = unnorm_h / norm_sum
    p_s = unnorm_s / norm_sum
    print(f"t{t} ({obs}): P(H_pred)={p_h_pred:.4f}, P(H_upd)={p_h:.4f}")


obs_sequence = ["ALARM", "NORMAL", "ALARM", "ALARM"]
hmm_filter(obs_sequence)
```

```python
import numpy as np

P_High = 0.5
P_Low = 0.5

P_Cold_High = 5 / 6
P_Rain_High = 4 / 6

P_Cold_Low = 2 / 6
P_Rain_Low = 1 / 6

Score_High = P_High * P_Cold_High * P_Rain_High
Score_Low = P_Low * P_Cold_Low * P_Rain_Low

Total_Score = Score_High + Score_Low
P_High_given_CR = Score_High / Total_Score
P_Low_given_CR = Score_Low / Total_Score

print(f"Score(High): {Score_High:.3f}")
print(f"Score(Low): {Score_Low:.3f}")
print(f"P(High | Cold, Rain): {P_High_given_CR:.2f}")
print(f"P(Low | Cold, Rain): {P_Low_given_CR:.2f}")

laplace_event_1 = (0 + 1) / (6 + 2)
laplace_event_0 = (6 + 1) / (6 + 2)

print(f"Лаплас тегістеуі P(Event = 1 | Low): {laplace_event_1:.3f}")
print(f"Лаплас тегістеуі P(Event = 0 | Low): {laplace_event_0:.3f}")
```
